### 2.3 Variabili
Una **variabile** è un contenitore con un nome che memorizza un valore da riutilizzare. Si crea con l'**assegnazione** `=`: a sinistra il nome, a destra il valore.

In Python non si dichiara il tipo: viene deciso dal valore.

**Regole per i nomi**
- Possono contenere lettere, numeri e underscore `_`, ma **non** iniziare con un numero.
- Non possono contenere spazi o caratteri speciali (`-`, `€`, `!`…).
- Non possono essere parole riservate (`if`, `for`, `print`…).
- Maiuscole e minuscole sono diverse: `eta` e `Eta` sono due variabili distinte.

**Convenzione:** `snake_case` (minuscolo, parole separate da `_`), ad esempio `altitudine_massima`.

````python
nome = "Giulia"
eta = 24
altezza = 1.68
maggiorenne = True

print(nome)
print(eta)
print(altezza)
print(maggiorenne)
````

Il valore di una variabile si può cambiare in qualsiasi momento: vale l'ultima assegnazione.

````python
punti = 10
print(punti)
punti = 25
print(punti)
punti = punti + 5   # prende il valore attuale, aggiunge 5, lo risalva
print(punti)
````

### 2.7 Operatori aritmetici
| Operatore | Operazione | Esempio | Risultato |
|---|---|---|---|
| `+` | addizione | `10 + 5` | `15` |
| `-` | sottrazione | `10 - 5` | `5` |
| `*` | moltiplicazione | `10 * 5` | `50` |
| `/` | divisione (sempre `float`) | `10 / 5` | `2.0` |
| `//` | divisione intera | `10 // 3` | `3` |
| `%` | modulo (resto) | `10 % 3` | `1` |
| `**` | potenza | `2 ** 3` | `8` |

**Precedenza:** prima `**`, poi `*` `/` `//` `%`, infine `+` `-`. Le parentesi `()` hanno la precedenza assoluta.

````python
x = 17
y = 5
print("Addizione:", x + y)
print("Sottrazione:", x - y)
print("Moltiplicazione:", x * y)
print("Divisione:", x / y)
print("Divisione intera:", x // y)
print("Resto:", x % y)
print("Potenza:", x ** 2)
print("Precedenza:", 2 + 3 * 4, "contro", (2 + 3) * 4)
````

**Operatori di assegnazione composta:** scorciatoie per aggiornare una variabile.

````python
n = 10
n += 3     # come n = n + 3
print(n)
n -= 1
n *= 2
print(n)
````

**Il modulo `%` nella pratica:** con `% 2` si capisce se un numero è pari (resto 0) o dispari (resto 1); con `// 60` e `% 60` si convertono i secondi in minuti.

````python
secondi_totali = 3725
minuti = secondi_totali // 60
secondi = secondi_totali % 60
print(minuti, "minuti e", secondi, "secondi")
````

### 3.2 Caratteri speciali e escape
Il carattere `\` (backslash) dà un significato speciale al carattere successivo.

| Sequenza | Significato |
|---|---|
| `\n` | a capo |
| `\t` | tabulazione |
| `\"` | virgolette doppie dentro un testo |
| `\'` | apostrofo dentro un testo con apici |
| `\\` | un backslash vero e proprio |

````python
print("Prima riga\nSeconda riga")
print("Colonna A\tColonna B")
print("Ha detto: \"Ciao, come va?\"")
print('L\'altro giorno')
print("C:\\Utenti\\Documenti\\file.txt")
````

**Alternativa:** per l'apostrofo si può usare le virgolette doppie come delimitatori: `"L'altro giorno"`.

### 3.3 Concatenazione e ripetizione
- `+` unisce due stringhe (non aggiunge spazi da sé!). Non si può unire un testo a un numero senza `str()`.
- `*` ripete una stringa.

````python
nome = "Marta"
cognome = "Rossi"
completo = nome + " " + cognome
print(completo)

separatore = "=" * 30
print(separatore)
print("Titolo")
print(separatore)
````

### 3.4 f-string
Una **f-string** si scrive con una `f` davanti alle virgolette. Dentro le parentesi graffe `{}` si inseriscono variabili o anche espressioni. È il modo più leggibile per costruire frasi.

````python
nome = "Marta"
eta = 28
citta = "Torino"

print(f"Mi chiamo {nome}, ho {eta} anni e vivo a {citta}.")
print(f"Tra 10 anni avrò {eta + 10} anni.")
print(f"Il mio nome ha {len(nome)} lettere.")
````

**Confronto con la concatenazione** (più lunga e piena di `str()`):

````python
print("Mi chiamo " + nome + ", ho " + str(eta) + " anni e vivo a " + citta + ".")
````

### 3.5 Formattazione dei numeri nelle f-string
Dopo i due punti `:` dentro le graffe si indica il formato.

| Formato | Effetto | Esempio |
|---|---|---|
| `{x:.2f}` | 2 decimali | `19.997` → `20.00` |
| `{x:.0%}` | percentuale | `0.15` → `15%` |
| `{x:,}` | separatore delle migliaia | `1234567` → `1,234,567` |
| `{x:>10}` | allinea a destra in 10 spazi | |
| `{x:<10}` | allinea a sinistra | |
| `{x:^10}` | centra | |
| `{x:05d}` | riempie con zeri fino a 5 cifre | `42` → `00042` |

````python
valore = 19.99674
sconto = 0.15
grande = 1234567
numero = 42

print(f"Valore con 2 decimali: {valore:.2f}")
print(f"Valore con 1 decimale: {valore:.1f}")
print(f"Lo sconto è del {sconto:.0%}")
print(f"Numero grande: {grande:,}")
print(f"Codice: {numero:05d}")
print(f"[{'ciao':>10}] [{'ciao':<10}] [{'ciao':^10}]")
````

## 4. Stringhe II: metodi e slicing

Un **metodo** è una funzione "legata" a un valore, si scrive con il punto: `testo.upper()`. ⚠️ Le stringhe sono **immutabili**: i metodi non modificano l'originale, **restituiscono una nuova stringa** che va salvata in una variabile se serve.

### 4.1 Maiuscole e minuscole

````python
testo = "hello world!"
print(testo.upper())        # TUTTO MAIUSCOLO
print(testo.lower())        # tutto minuscolo
print(testo.capitalize())   # solo la prima lettera maiuscola
print(testo.title())        # ogni parola con l'iniziale maiuscola
print(testo)                # l'originale non è cambiato!
````

### 4.2 Rimuovere gli spazi
`strip()` toglie gli spazi a inizio e fine (`lstrip` solo a sinistra, `rstrip` solo a destra). Utilissimo per ripulire l'input dell'utente.

````python
testo = "   testo con spazi   "
print(f"[{testo}]")
print(f"[{testo.strip()}]")
print(f"[{testo.lstrip()}]")
print(f"[{testo.rstrip()}]")
````

### 4.3 Sostituire
`replace(vecchio, nuovo)` sostituisce tutte le occorrenze; un terzo argomento limita il numero di sostituzioni.

````python
frase = "Python è bello. Python è utile."
print(frase.replace("Python", "Java"))
print(frase.replace("Python", "Java", 1))
````

### 4.4 Cercare e contare
- `find(x)` dà la posizione della **prima** occorrenza, o `-1` se non c'è. `rfind(x)` parte da destra.
- `count(x)` conta le occorrenze (distingue maiuscole e minuscole: per ignorarle si usa prima `.lower()`).
- `in` verifica se un testo è contenuto in un altro.

````python
testo = "Mi piace la pizza, la Pizza è buonissima"
print(testo.find("pizza"))
print(testo.find("pasta"))
print(testo.rfind("la"))
print(testo.count("pizza"))
print(testo.lower().count("pizza"))
print("pizza" in testo)
````

### 4.5 Inizio e fine

### 4.6 Dividere e unire
- `split()` senza argomenti divide dove trova spazi (anche multipli) e restituisce una **lista** di parole.
- `split(sep)` divide usando un separatore a scelta.
- `separatore.join(lista)` fa l'operazione inversa.

````python
frase = "Sono   bravo a   cucinare"
parole = frase.split()
print(parole)
print(" ".join(parole))          # spazi normalizzati

elenco = "frutta,verdura,legumi"
print(elenco.split(","))
````

### 4.7 Verificare il contenuto
Questi metodi restituiscono `True` o `False`.

### 4.9 Slicing
Ogni carattere ha una **posizione (indice)**, che parte da **0**. Con indici negativi si conta da destra: `-1` è l'ultimo.

```
 testo =  P  y  t  h  o  n
 indice:  0  1  2  3  4  5
 negativo:-6 -5 -4 -3 -2 -1
```
Sintassi: `testo[inizio:fine:passo]` — l'inizio è **incluso**, la fine è **esclusa**.

````python
testo = "Python"
print(testo[0])        # primo carattere
print(testo[-1])       # ultimo carattere
print(testo[0:3])      # Pyt
print(testo[:3])       # come sopra
print(testo[3:])       # hon
print(testo[-3:])      # ultimi 3 caratteri
print(testo[::2])      # un carattere ogni due: Pto
print(testo[::-1])     # al contrario: nohtyP
````

Slicing e `find` insieme permettono di "tagliare" pezzi di testo, ad esempio una email.

````python
email = "mario.rossi@esempio.it"
posizione_chiocciola = email.find("@")
utente = email[:posizione_chiocciola]
dominio = email[posizione_chiocciola + 1:]
print(utente)
print(dominio)
````
