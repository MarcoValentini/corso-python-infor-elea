## 2. Git e GitHub

### 2.1 A cosa serve
**Git** è un programma che registra la **cronologia** dei file di un progetto: puoi tornare a una versione precedente, vedere cosa è cambiato e chi l'ha cambiato. **GitHub** è un sito che ospita copie online dei progetti Git e permette di condividerli e lavorare in gruppo.

| Termine | Significato |
|---|---|
| **Repository** (repo) | cartella di progetto tracciata da Git |
| **Commit** | "fotografia" salvata della cartella, con un messaggio descrittivo |
| **Staging** | area d'attesa dove si scelgono le modifiche da includere nel prossimo commit |
| **Locale / remoto** | copia sul tuo computer / copia su GitHub |
| **Push** | invia i commit locali a GitHub |

**Il ciclo di lavoro:** modifichi i file → `git add` (scegli cosa salvare) → `git commit` (salvi) → `git push` (invii online).

```
cartella di lavoro --git add--> staging --git commit--> repository locale --git push--> GitHub
```

### 2.4 Il primo repository
Esegui questi comandi **uno alla volta** nel terminale e leggi sempre cosa risponde Git.
```
mkdir esercizi-python
cd esercizi-python
git init
```
`git init` crea il repository (una cartella nascosta `.git`). Ora crea un file `ciao.py` con una sola riga: `print("Ciao, mondo!")` e prosegui:
```
git status
git add ciao.py
git status
git commit -m "Aggiunge il primo script"
git log --oneline
```
- `git status` mostra lo stato: file non tracciati, modificati, pronti per il commit. **È il comando da usare più spesso.**
- Il messaggio del commit (`-m`) descrive **cosa** hai cambiato, in poche parole ("Aggiunge calcolo del carburante", non "modifiche").
- `git log --oneline` mostra la cronologia dei commit.

Ora modifica `ciao.py` (aggiungi un'altra `print`) e ripeti il ciclo:
```
git diff
git add ciao.py
git commit -m "Aggiunge un secondo messaggio"
git log --oneline
```
`git diff` mostra le righe cambiate (in verde le aggiunte, in rosso le rimozioni) prima di salvarle.

### 3.3 Variabili
Una **variabile** è un contenitore con un nome che memorizza un valore da riutilizzare. Si crea con l'**assegnazione** `=`: a sinistra il nome, a destra il valore.

In Python non si dichiara il tipo: viene deciso dal valore.

**Regole per i nomi**
- Possono contenere lettere, numeri e underscore `_`, ma **non** iniziare con un numero.
- Non possono contenere spazi o caratteri speciali (`-`, `€`, `!`…).
- Non possono essere parole riservate (`if`, `for`, `print`…).
- Maiuscole e minuscole sono diverse: `eta` e `Eta` sono due variabili distinte.

**Convenzione:** `snake_case` (minuscolo, parole separate da `_`), ad esempio `altitudine_massima`.

````python
nome = "Giulia"
eta = 24
altezza = 1.68
maggiorenne = True

print(nome)
print(eta)
print(altezza)
print(maggiorenne)
````

Il valore di una variabile si può cambiare in qualsiasi momento: vale l'ultima assegnazione.

````python
punti = 10
print(punti)
punti = 25
print(punti)
punti = punti + 5   # prende il valore attuale, aggiunge 5, lo risalva
print(punti)
````

### 3.5 Conversione di tipo
Si converte un valore con `str()`, `int()`, `float()`, `bool()`.

⚠️ Alcune conversioni falliscono se il valore non è compatibile (`int("ciao")` genera un errore). Inoltre:
- `int(3.7)` **tronca** la parte decimale: dà `3`, non `4`.
- `bool("")` è `False` solo per la stringa vuota; `bool("False")` e `bool("0")` sono `True`, perché sono stringhe non vuote.

````python
eta = 22
frase = "Ho " + str(eta) + " anni"       # numero -> testo, per poterlo unire
print(frase)

testo = "22"
print(int(testo) + 5)                     # testo -> numero, per poterci calcolare

testo_decimale = "3.14"
print(float(testo_decimale) + 5)

print(int(3.7))                           # tronca, non arrotonda

print(bool(""), bool("False"), bool("0"), bool(0), bool(5))
````

### 3.7 Operatori aritmetici
| Operatore | Operazione | Esempio | Risultato |
|---|---|---|---|
| `+` | addizione | `10 + 5` | `15` |
| `-` | sottrazione | `10 - 5` | `5` |
| `*` | moltiplicazione | `10 * 5` | `50` |
| `/` | divisione (sempre `float`) | `10 / 5` | `2.0` |
| `//` | divisione intera | `10 // 3` | `3` |
| `%` | modulo (resto) | `10 % 3` | `1` |
| `**` | potenza | `2 ** 3` | `8` |

**Precedenza:** prima `**`, poi `*` `/` `//` `%`, infine `+` `-`. Le parentesi `()` hanno la precedenza assoluta.

````python
x = 17
y = 5
print("Addizione:", x + y)
print("Sottrazione:", x - y)
print("Moltiplicazione:", x * y)
print("Divisione:", x / y)
print("Divisione intera:", x // y)
print("Resto:", x % y)
print("Potenza:", x ** 2)
print("Precedenza:", 2 + 3 * 4, "contro", (2 + 3) * 4)
````

**Operatori di assegnazione composta:** scorciatoie per aggiornare una variabile.

````python
n = 10
n += 3     # come n = n + 3
print(n)
n -= 1
n *= 2
print(n)
````

**Il modulo `%` nella pratica:** con `% 2` si capisce se un numero è pari (resto 0) o dispari (resto 1); con `// 60` e `% 60` si convertono i secondi in minuti.

````python
secondi_totali = 3725
minuti = secondi_totali // 60
secondi = secondi_totali % 60
print(minuti, "minuti e", secondi, "secondi")
````

## 4. Stringhe I: creazione, escape, f-string, formattazione

Una **stringa** è una sequenza di caratteri (testo). Si scrive tra virgolette doppie `"..."` o singole `'...'`. Con tre virgolette `"""..."""` si scrivono testi su più righe.

````python
nome = "Marta"
cognome = 'Rossi'

descrizione = """Questo è un testo
scritto su più righe"""
print(descrizione)
````

### 4.1 Lunghezza
`len()` restituisce il numero di caratteri (spazi inclusi).

````python
testo = "Hello, World!"
print("Lunghezza:", len(testo))
````

### 4.2 Caratteri speciali e escape
Il carattere `\` (backslash) dà un significato speciale al carattere successivo.

| Sequenza | Significato |
|---|---|
| `\n` | a capo |
| `\t` | tabulazione |
| `\"` | virgolette doppie dentro un testo |
| `\'` | apostrofo dentro un testo con apici |
| `\\` | un backslash vero e proprio |

````python
print("Prima riga\nSeconda riga")
print("Colonna A\tColonna B")
print("Ha detto: \"Ciao, come va?\"")
print('L\'altro giorno')
print("C:\\Utenti\\Documenti\\file.txt")
````

**Alternativa:** per l'apostrofo si può usare le virgolette doppie come delimitatori: `"L'altro giorno"`.

### 4.3 Concatenazione e ripetizione
- `+` unisce due stringhe (non aggiunge spazi da sé!). Non si può unire un testo a un numero senza `str()`.
- `*` ripete una stringa.

````python
nome = "Marta"
cognome = "Rossi"
completo = nome + " " + cognome
print(completo)

separatore = "=" * 30
print(separatore)
print("Titolo")
print(separatore)
````

### 4.4 f-string
Una **f-string** si scrive con una `f` davanti alle virgolette. Dentro le parentesi graffe `{}` si inseriscono variabili o anche espressioni. È il modo più leggibile per costruire frasi.

````python
nome = "Marta"
eta = 28
citta = "Torino"

print(f"Mi chiamo {nome}, ho {eta} anni e vivo a {citta}.")
print(f"Tra 10 anni avrò {eta + 10} anni.")
print(f"Il mio nome ha {len(nome)} lettere.")
````

**Confronto con la concatenazione** (più lunga e piena di `str()`):

````python
print("Mi chiamo " + nome + ", ho " + str(eta) + " anni e vivo a " + citta + ".")
````

### 4.5 Formattazione dei numeri nelle f-string
Dopo i due punti `:` dentro le graffe si indica il formato.

| Formato | Effetto | Esempio |
|---|---|---|
| `{x:.2f}` | 2 decimali | `19.997` → `20.00` |
| `{x:.0%}` | percentuale | `0.15` → `15%` |
| `{x:,}` | separatore delle migliaia | `1234567` → `1,234,567` |
| `{x:>10}` | allinea a destra in 10 spazi | |
| `{x:<10}` | allinea a sinistra | |
| `{x:^10}` | centra | |
| `{x:05d}` | riempie con zeri fino a 5 cifre | `42` → `00042` |

````python
valore = 19.99674
sconto = 0.15
grande = 1234567
numero = 42

print(f"Valore con 2 decimali: {valore:.2f}")
print(f"Valore con 1 decimale: {valore:.1f}")
print(f"Lo sconto è del {sconto:.0%}")
print(f"Numero grande: {grande:,}")
print(f"Codice: {numero:05d}")
print(f"[{'ciao':>10}] [{'ciao':<10}] [{'ciao':^10}]")
````
