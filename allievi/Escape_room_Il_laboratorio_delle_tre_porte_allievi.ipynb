## Test di metà corso: «Escape room — Il laboratorio delle tre porte»

Un esercizio individuale da circa **60 minuti**, per ripassare variabili, input, conversioni, stringhe, calcoli, confronti, operatori logici, `if/elif/else` e operatore ternario. Non richiede cicli, liste o funzioni.

La consegna seguente è pronta da incollare in una cella Markdown di Jupyter. Gli allievi possono svolgere ogni fase in una cella di codice diversa, eseguendole in ordine.

# Escape room: il laboratorio delle tre porte

Ti risvegli in un laboratorio abbandonato. Una voce metallica annuncia:

> «La buona notizia: esiste un'uscita.
> La cattiva notizia: chi ha progettato il sistema di sicurezza aveva appena scoperto Python.»

Per uscire devi superare tre porte. Ogni prova assegna punti, ma nessuna ti impedisce di affrontare la successiva: anche se sbagli, continua il programma.

## Regole del gioco

- Usa soltanto gli argomenti già affrontati.
- Non usare cicli, liste, funzioni o librerie esterne.
- Usa almeno una f-string.
- Usa almeno un confronto concatenato, per esempio `10 <= valore <= 20`.
- Usa almeno una condizione con `and`, una con `or` e una con `not`.
- Usa almeno un operatore ternario.
- Il programma deve terminare mostrando un rapporto finale.

Per le domande numeriche puoi supporre che l'utente inserisca un numero valido: non devi gestire il caso in cui scriva lettere.

## Fase 1 — Il badge dell'esploratore

Il terminale chiede:

1. Il tuo nome.
2. Un nome in codice.

Ripulisci entrambi dagli spazi iniziali e finali:

- Il nome deve avere l'iniziale maiuscola e le altre lettere minuscole.
- Il nome in codice deve diventare tutto maiuscolo.

Un nome in codice è valido se:

- contiene da 4 a 10 caratteri inclusi;
- contiene soltanto lettere o numeri;
- non è composto esclusivamente da cifre.

Esempi:

| Nome in codice | Valido? | Motivo |
|---|---|---|
| `VOLPE7` | Sì | Rispetta tutte le condizioni |
| `12345` | No | Contiene soltanto cifre |
| `AI` | No | È troppo corto |
| `SUPER EROE` | No | Contiene uno spazio |

Crea una variabile `punti` con valore iniziale `0`.

- Se il nome in codice è valido, assegna 10 punti e stampa un messaggio di benvenuto.
- Altrimenti, non assegnare punti e stampa: `Badge sospetto. Ti chiameremo OSPITE.`
- In questo secondo caso, sostituisci il nome in codice con `"OSPITE"`.

Suggerimenti: `strip()`, `capitalize()`, `upper()`, `len()`, `isalnum()`, `isdigit()`.

## Fase 2 — La porta del codice segreto

Sulla porta compare questo messaggio:

> «Prendi il numero 47.
> Scrivi prima il quoziente della divisione intera per 6.
> Subito dopo scrivi il resto della stessa divisione.
> Quello è il codice.»

Il programma deve:

1. Calcolare il codice usando `//` e `%`.
2. Chiedere all'utente di inserirlo.
3. Confrontare la risposta con il codice calcolato.

Punteggio:

- Codice corretto: 25 punti.
- Codice sbagliato: 0 punti.

Stampa un messaggio diverso nei due casi.

Attenzione: il codice è un testo formato dall'unione di due risultati, non dalla loro somma.

## Fase 3 — La porta della pozione instabile

Trovi una pozione che può alimentare il meccanismo della porta. Il terminale chiede:

- la temperatura della pozione, come numero decimale;
- il colore della pozione;
- se ci sono bolle: l'utente risponde `sì` oppure `no`.

Ripulisci le risposte testuali e convertile in minuscolo, così `" VERDE "` e `"verde"` sono equivalenti.

La pozione è sicura soltanto se:

- la temperatura è compresa tra 18 e 25 gradi inclusi;
- il colore è `"verde"` oppure `"blu"`;
- la risposta alla domanda sulle bolle è `"no"`.

Punteggio:

- Pozione sicura: 25 punti.
- Pozione non sicura: 0 punti.

Stampa un messaggio appropriato, per esempio:

```text
Pozione stabile: la porta si apre con un discreto rumore di lavatrice.
```

oppure:

```text
Pozione instabile: una pianta nelle vicinanze ha appena imparato a parlare.
```

Per verificare la temperatura usa un confronto concatenato.

## Fase 4 — La porta del distributore automatico

L'ultima porta ha un distributore che accetta soltanto monete da 2 euro. Devi acquistare una batteria d'emergenza.

Chiedi:

1. Quante monete da 2 euro inserisce l'utente.
2. Quale batteria sceglie: `"mini"`, `"normale"` o `"mega"`.

Il listino è:

| Batteria | Prezzo |
|---|---:|
| mini | 4 euro |
| normale | 6 euro |
| mega | 10 euro |

Il programma deve:

1. Calcolare il denaro inserito.
2. Identificare il prezzo della batteria con `if/elif/else`.
3. Controllare se il denaro è sufficiente.
4. In caso di acquisto riuscito, calcolare il resto.

Punteggio:

- Batteria riconosciuta e denaro sufficiente: 30 punti.
- Batteria riconosciuta ma denaro insufficiente: 10 punti.
- Batteria non riconosciuta: 0 punti.

Se la batteria non esiste, stampa:

```text
Batteria sconosciuta. Il distributore ti restituisce uno sguardo deluso.
```

Se il denaro è insufficiente, indica quanti euro mancano.

Se l'acquisto riesce, indica batteria acquistata e resto.

Il numero di monete inserito sarà un intero maggiore o uguale a zero.

## Fase 5 — Il rapporto di fuga

Il punteggio massimo è 90.

Assegna un titolo finale:

| Punti | Titolo |
|---|---|
| Da 75 a 90 | Maestro della fuga |
| Da 50 a 74 | Esploratore intraprendente |
| Da 25 a 49 | Sopravvissuto per caso |
| Da 0 a 24 | Nuovo arredamento del laboratorio |

Usa inoltre un operatore ternario per assegnare alla variabile `esito`:

- `"USCITA SBLOCCATA"` se i punti sono almeno 50;
- `"USCITA BLOCCATA"` altrimenti.

Stampa un rapporto simile:

```text
========================================
RAPPORTO DEL LABORATORIO
Esploratore: Marta
Nome in codice: VOLPE7
Punteggio: 90/90
Titolo: Maestro della fuga
Esito: USCITA SBLOCCATA
========================================
```

Usa f-string e una stringa ripetuta con `*` per costruire i separatori.
