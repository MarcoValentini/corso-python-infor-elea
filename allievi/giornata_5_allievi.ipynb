## 1. File di testo

Finora i dati vivevano solo in memoria: chiuso il programma, sparivano. Con i **file** i dati si salvano sul disco e si possono rileggere in seguito.

### 1.1 Scrivere un file
`open(nome, modalità)` apre un file. La modalità `"w"` (*write*) lo crea, oppure lo **svuota se esiste già**. Con `write()` si scrive del testo; `\n` serve per andare a capo. Alla fine il file va **chiuso** con `close()`.

````python
file = open("esempio.txt", "w", encoding="utf-8")
file.write("Ciao, mondo!")
file.write("\nSeconda riga")
file.close()
````

Controlla che nella cartella sia comparso `esempio.txt`.

### 1.2 Leggere un file
Modalità `"r"` (*read*). Esistono più modi di leggere:
- `read()`: tutto il contenuto in un'unica stringa;
- ciclo `for`: una riga alla volta;
- `readlines()`: lista con tutte le righe;
- `readline()`: una sola riga per volta.

Ogni riga letta termina con `\n`: si toglie con `strip()`.

````python
file = open("esempio.txt", "r", encoding="utf-8")
contenuto = file.read()
print(contenuto)
file.close()
````

````python
file = open("esempio.txt", "r", encoding="utf-8")
for riga in file:
    print(riga.strip())
file.close()
````

````python
file = open("esempio.txt", "r", encoding="utf-8")
righe = file.readlines()
print(righe)                 # ['Ciao, mondo!\n', 'Seconda riga']
file.close()

file = open("esempio.txt", "r", encoding="utf-8")
print(file.readline().strip())     # solo la prima
print(file.readline().strip())     # poi la seconda
file.close()
````

### 1.3 `with`: la forma consigliata
Dimenticare `close()` è facile. L'istruzione `with` chiude il file **automaticamente**, anche se nel blocco c'è un errore. Da qui in poi useremo sempre questa forma.

````python
with open("esempio.txt", "r", encoding="utf-8") as file:
    contenuto = file.read()

print(contenuto)     # il file è già chiuso, ma la variabile contenuto resta
````

### 1.4 Aggiungere in coda: modalità `"a"`
`"a"` (*append*) scrive **in fondo** senza cancellare ciò che c'è.

````python
with open("esempio.txt", "a", encoding="utf-8") as file:
    file.write("\nNuova riga in coda")

with open("esempio.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

### 1.5 Le modalità principali
| Modalità | Significato | Se il file esiste | Se non esiste |
|---|---|---|---|
| `"r"` | lettura | lo legge | **errore** `FileNotFoundError` |
| `"w"` | scrittura | **lo svuota** | lo crea |
| `"a"` | aggiunta in coda | aggiunge in fondo | lo crea |
| `"x"` | creazione esclusiva | **errore** `FileExistsError` | lo crea |

⚠️ La più pericolosa è `"w"`: cancella tutto senza avvisare.

### 1.6 Codifica dei caratteri
Indica sempre `encoding="utf-8"`: così lettere accentate e simboli (è, à, €) si salvano e si leggono correttamente su Windows, macOS e Linux.

### 1.7 File che non esiste: gestire l'errore
Si combina con `try`/`except` (giornata 3):

````python
try:
    with open("non_esiste.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("Il file non esiste!")
````

### 1.8 (facoltativo) Il puntatore: `tell()` e `seek()`
Dopo una lettura, Python "si ricorda" a che punto del file è arrivato. `tell()` dice la posizione, `seek(n)` si sposta al carattere `n`.

````python
with open("esempio.txt", "r", encoding="utf-8") as file:
    print(file.tell())          # 0: all'inizio
    print(file.read(5))         # legge 5 caratteri
    print(file.tell())          # 5
    file.seek(0)                # torna all'inizio
    print(file.readline().strip())
````

## 2. Il formato JSON

**JSON** (*JavaScript Object Notation*) è un formato di testo per memorizzare e scambiare dati strutturati. È lo standard usato da siti e servizi web per scambiarsi informazioni: lo vedrai spesso.

Somiglia molto a dizionari e liste di Python:
```
{
  "nome": "Anna",
  "anni": 31,
  "hobby": ["lettura", "corsa"],
  "iscritta": true,
  "indirizzo": null
}
```

### 2.1 Corrispondenza Python ↔ JSON
| Python | JSON |
|---|---|
| `dict` | oggetto `{...}` |
| `list`, `tuple` | array `[...]` |
| `str` | stringa |
| `int`, `float` | numero |
| `True` / `False` | `true` / `false` |
| `None` | `null` |

Si usa il modulo `json`.

````python
import json
````

### 2.2 Da Python a testo JSON: `json.dumps`
`dumps` (*dump string*) converte un dizionario o una lista in una **stringa** JSON. `indent=2` la rende leggibile.

````python
utente = {"nome": "Anna", "anni": 31, "abilitata": True, "indirizzo": None, "hobby": ["lettura", "corsa"]}

testo_json = json.dumps(utente, indent=2)
print(testo_json)
print(type(utente), type(testo_json))
````

### 2.3 Da testo JSON a Python: `json.loads`
`loads` (*load string*) fa l'operazione inversa.

````python
testo = '{"nome": "Luca", "anni": 25, "abilitato": true}'
dati = json.loads(testo)
print(dati)
print(dati["nome"], type(dati))
````

### 2.4 Scrivere e leggere file JSON: `dump` e `load`
Senza la `s` finale lavorano direttamente su un **file**: `json.dump(dati, file)` e `json.load(file)`.

`ensure_ascii=False` permette di salvare le lettere accentate così come sono.

````python
utente = {"nome_utente": "anna.verdi", "email": "anna@example.com", "anni": 31, "abilitata": True}

with open("utente.json", "w", encoding="utf-8") as file:
    json.dump(utente, file, indent=2, ensure_ascii=False)

with open("utente.json", "r", encoding="utf-8") as file:
    letto = json.load(file)

print(letto)
print(letto["email"])
````

⚠️ **Attenzione alle chiavi:** in JSON le chiavi sono **sempre testo**. Una chiave numerica diventa una stringa dopo il salvataggio.

````python
print(json.dumps({1: "uno", 2: "due"}))      # le chiavi diventano "1" e "2"
print(json.loads(json.dumps({1: "uno"})))    # tornano come stringhe
````

## 3. Il formato CSV

**CSV** (*Comma-Separated Values*) rappresenta una **tabella** come testo: una riga per record, valori separati da un delimitatore (più comuni: `,` `;` tab). La prima riga contiene di solito i nomi delle colonne (**intestazione**).
```
codice,distanza_km,passeggeri
AZ101,650,120
LH202,1200,85
```
Si può aprire con Excel o LibreOffice. Si usa il modulo `csv`.

````python
import csv
````

### 3.1 Scrivere un CSV
`csv.writer` scrive le righe da liste. `writerow` scrive una riga, `writerows` molte. 

⚠️ Apri sempre il file con `newline=""` per evitare righe vuote in più (soprattutto su Windows).

````python
tabella = [
    ["codice", "distanza_km", "passeggeri"],
    ["AZ101", 650, 120],
    ["LH202", 1200, 85],
]

with open("voli.csv", "w", encoding="utf-8", newline="") as file:
    scrittore = csv.writer(file, delimiter=",")
    scrittore.writerows(tabella)
    scrittore.writerow(["AF303", 480, 150])    # una riga in più
````

### 3.2 Leggere un CSV
`csv.reader` restituisce ogni riga come **lista**. Con `next()` si estrae l'intestazione.

⚠️ **Tutti i valori letti sono testo**: per calcolare vanno convertiti con `int()` o `float()`.

````python
with open("voli.csv", "r", encoding="utf-8", newline="") as file:
    lettore = csv.reader(file, delimiter=",")
    intestazione = next(lettore)
    righe = list(lettore)

print(intestazione)
print(righe)

totale = 0
for riga in righe:
    totale += int(riga[2])        # conversione da testo a numero
print("Passeggeri totali:", totale)
````

### 3.3 `DictReader` e `DictWriter`: righe come dizionari
Più comodo e leggibile: ogni riga diventa un **dizionario** con i nomi delle colonne come chiavi, così non serve ricordare gli indici.

````python
with open("voli.csv", "r", encoding="utf-8", newline="") as file:
    voli = list(csv.DictReader(file))

print(voli[0])
for volo in voli:
    print(volo["codice"], int(volo["passeggeri"]))
````

````python
righe = [
    {"codice": "BA404", "distanza_km": 1800, "passeggeri": 60},
    {"codice": "IB505", "distanza_km": 950, "passeggeri": 110},
]

with open("voli2.csv", "w", encoding="utf-8", newline="") as file:
    scrittore = csv.DictWriter(file, fieldnames=["codice", "distanza_km", "passeggeri"])
    scrittore.writeheader()
    scrittore.writerows(righe)
````

## 4. `pathlib` e `datetime`

### 4.1 `pathlib`: percorsi di file e cartelle
Un **percorso** indica dove si trova un file. Windows usa `\`, macOS e Linux usano `/`: con `pathlib` si scrive codice che funziona ovunque. L'operatore `/` unisce i pezzi del percorso.

````python
from pathlib import Path

cartella_corrente = Path.cwd()
print(cartella_corrente)

percorso = Path("dati") / "report" / "voli.txt"      # funziona su qualsiasi sistema
print(percorso)
print(percorso.name)       # voli.txt
print(percorso.stem)       # voli
print(percorso.suffix)     # .txt
print(percorso.parent)     # dati/report
````

### 4.2 `datetime`: date e orari
Il modulo `datetime` gestisce date e orari. Le classi principali:
- `date`: solo la data; `datetime`: data e ora; `timedelta`: una **durata** (differenza tra due momenti).

````python
from datetime import date, datetime, timedelta

oggi = date.today()
adesso = datetime.now()
print(oggi)
print(adesso)
print(oggi.year, oggi.month, oggi.day)

compleanno = date(1998, 5, 14)       # una data costruita a mano: anno, mese, giorno
print(compleanno)
````

**Calcoli con le date:** la differenza tra due date è un `timedelta`; si può sommare o sottrarre una durata.

````python
inizio = date(2026, 10, 21)
fine = date(2026, 11, 19)
differenza = fine - inizio
print(differenza.days, "giorni")

tra_trenta_giorni = inizio + timedelta(days=30)
print(tra_trenta_giorni)

print(date(2026, 11, 19).weekday())    # 0 = lunedì ... 6 = domenica
````

### 4.3 Da data a testo e viceversa
- `strftime(formato)`: data → testo formattato.
- `datetime.strptime(testo, formato)`: testo → data.

| Codice | Significato | Esempio |
|---|---|---|
| `%d` | giorno (2 cifre) | `07` |
| `%m` | mese (2 cifre) | `10` |
| `%Y` | anno (4 cifre) | `2026` |
| `%H:%M` | ore e minuti | `14:30` |
| `%S` | secondi | `05` |

````python
momento = datetime(2026, 11, 19, 14, 30)
print(momento.strftime("%d/%m/%Y"))
print(momento.strftime("%d/%m/%Y alle %H:%M"))

testo = "21/10/2026"
data = datetime.strptime(testo, "%d/%m/%Y")
print(data, type(data))
print(data.year)
````

## 5. Classi: introduzione

Finora abbiamo tenuto i dati in variabili e dizionari e il comportamento in funzioni separate. Una **classe** riunisce **dati** e **funzioni che li usano** in un unico "stampo". Con lo stampo si creano tanti **oggetti** (le **istanze**), ognuno con i propri dati.

Esempio concettuale: la classe `Volo` è lo stampo; `AZ101` e `LH202` sono due oggetti costruiti con quello stampo.

### 5.1 La prima classe
- `class Nome:` definisce la classe (per convenzione con iniziale maiuscola, `CamelCase`).
- `__init__` è il **costruttore**: viene eseguito quando si crea l'oggetto e imposta i dati iniziali (gli **attributi**).
- `self` è l'oggetto stesso: nei metodi serve per accedere ai suoi attributi (`self.nome`). Va sempre messo come primo parametro dei metodi, ma **non** si passa nella chiamata.
- Un **metodo** è una funzione dentro la classe.

````python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome          # attributo
        self.eta = eta

    def presenta(self):           # metodo
        print(f"Ciao, sono {self.nome} e ho {self.eta} anni.")

anna = Persona("Anna", 31)        # si crea un oggetto: __init__ viene eseguito
luca = Persona("Luca", 25)

anna.presenta()
luca.presenta()
print(anna.nome, luca.eta)
````

Ogni oggetto ha i **propri** attributi: modificare `anna` non tocca `luca`.

````python
anna.eta = 32
print(anna.eta, luca.eta)
````

### 5.2 Metodi che calcolano e metodi che modificano
Un metodo può restituire un valore calcolato dagli attributi, oppure **modificare lo stato** dell'oggetto.

````python
class ContoBancario:
    def __init__(self, titolare, saldo=0):
        self.titolare = titolare
        self.saldo = saldo

    def deposita(self, importo):
        self.saldo += importo

    def preleva(self, importo):
        if importo > self.saldo:
            print("Fondi insufficienti!")
        else:
            self.saldo -= importo

conto = ContoBancario("Anna", 100)
conto.deposita(50)
conto.preleva(30)
conto.preleva(500)
print(conto.saldo)
````

### 5.3 `__str__`: come si stampa un oggetto
Senza istruzioni, `print(oggetto)` mostra un indirizzo poco utile. Il metodo speciale `__str__` (nome fisso, con due underscore ai lati) decide cosa far stampare: deve **restituire** una stringa.

````python
class Volo:
    def __init__(self, codice, distanza_km, passeggeri):
        self.codice = codice
        self.distanza_km = distanza_km
        self.passeggeri = passeggeri

    def durata_ore(self, velocita_kmh=800):
        return self.distanza_km / velocita_kmh

    def __str__(self):
        return f"Volo {self.codice}: {self.distanza_km} km, {self.passeggeri} passeggeri"

v = Volo("AZ101", 650, 120)
print(v)
print(round(v.durata_ore(), 2))
````

### 5.4 Liste di oggetti
Gli oggetti si mettono in liste come qualsiasi altro valore, e si usano con comprehension, `sorted` e lambda.

````python
voli = [Volo("AZ101", 650, 120), Volo("LH202", 1200, 85), Volo("AF303", 480, 150)]

lunghi = [v.codice for v in voli if v.distanza_km > 600]
print(lunghi)

for v in sorted(voli, key=lambda v: v.passeggeri, reverse=True):
    print(v)
````

### 5.5 (facoltativo) Ereditarietà: un cenno
Una classe può **ereditare** da un'altra: riceve attributi e metodi e ne aggiunge di propri. Si indica tra parentesi il nome della classe "madre"; `super().__init__(...)` richiama il costruttore della madre.

````python
class Veicolo:
    def __init__(self, nome, velocita_kmh):
        self.nome = nome
        self.velocita_kmh = velocita_kmh

    def tempo_ore(self, distanza_km):
        return distanza_km / self.velocita_kmh

class Aereo(Veicolo):
    def __init__(self, nome, velocita_kmh, quota_piedi):
        super().__init__(nome, velocita_kmh)
        self.quota_piedi = quota_piedi

a = Aereo("A320", 830, 37000)
print(a.nome, a.tempo_ore(1660), a.quota_piedi)
````
