## 1. Stringhe II: metodi e slicing
Un **metodo** è una funzione "legata" a un valore, si scrive con il punto: `testo.upper()`. ⚠️ Le stringhe sono **immutabili**: i metodi non modificano l'originale, **restituiscono una nuova stringa** che va salvata in una variabile se serve.

### 1.1 Maiuscole e minuscole

````python
testo = "hello world!"
print(testo.upper())        # TUTTO MAIUSCOLO
print(testo.lower())        # tutto minuscolo
print(testo.capitalize())   # solo la prima lettera maiuscola
print(testo.title())        # ogni parola con l'iniziale maiuscola
print(testo)                # l'originale non è cambiato!
````

### 1.2 Rimuovere gli spazi
`strip()` toglie gli spazi a inizio e fine (`lstrip` solo a sinistra, `rstrip` solo a destra). Utilissimo per ripulire l'input dell'utente.

````python
testo = "   testo con spazi   "
print(f"[{testo}]")
print(f"[{testo.strip()}]")
print(f"[{testo.lstrip()}]")
print(f"[{testo.rstrip()}]")
````

### 1.3 Sostituire
`replace(vecchio, nuovo)` sostituisce tutte le occorrenze; un terzo argomento limita il numero di sostituzioni.

````python
frase = "Python è bello. Python è utile."
print(frase.replace("Python", "Java"))
print(frase.replace("Python", "Java", 1))
````

### 1.4 Cercare e contare
- `find(x)` dà la posizione della **prima** occorrenza, o `-1` se non c'è. `rfind(x)` parte da destra.
- `count(x)` conta le occorrenze (distingue maiuscole e minuscole: per ignorarle si usa prima `.lower()`).
- `in` verifica se un testo è contenuto in un altro.

````python
testo = "Mi piace la pizza, la Pizza è buonissima"
print(testo.find("pizza"))
print(testo.find("pasta"))
print(testo.rfind("la"))
print(testo.count("pizza"))
print(testo.lower().count("pizza"))
print("pizza" in testo)
````

### 1.5 Inizio e fine

````python
nome_file = "documento.pdf"
print(nome_file.startswith("doc"))
print(nome_file.endswith(".pdf"))
````

### 1.6 Dividere e unire
- `split()` senza argomenti divide dove trova spazi (anche multipli) e restituisce una **lista** di parole.
- `split(sep)` divide usando un separatore a scelta.
- `separatore.join(lista)` fa l'operazione inversa.

````python
frase = "Sono   bravo a   cucinare"
parole = frase.split()
print(parole)
print(" ".join(parole))          # spazi normalizzati

elenco = "frutta,verdura,legumi"
print(elenco.split(","))
````

### 1.7 Verificare il contenuto
Questi metodi restituiscono `True` o `False`.

````python
print("12345".isdigit())       # solo cifre?
print("Python".isalpha())      # solo lettere?
print("Python3".isalnum())     # lettere e/o cifre?
print("CIAO".isupper(), "ciao".islower())
````

### 1.8 Allineamento
`center`, `ljust`, `rjust` riempiono una stringa fino a una larghezza data.

````python
titolo = "Python"
print(titolo.center(30, "-"))
print(titolo.ljust(30, "."))
print(titolo.rjust(30, "."))
````

### 1.9 Slicing
Ogni carattere ha una **posizione (indice)**, che parte da **0**. Con indici negativi si conta da destra: `-1` è l'ultimo.

```
 testo =  P  y  t  h  o  n
 indice:  0  1  2  3  4  5
 negativo:-6 -5 -4 -3 -2 -1
```
Sintassi: `testo[inizio:fine:passo]` — l'inizio è **incluso**, la fine è **esclusa**.

````python
testo = "Python"
print(testo[0])        # primo carattere
print(testo[-1])       # ultimo carattere
print(testo[0:3])      # Pyt
print(testo[:3])       # come sopra
print(testo[3:])       # hon
print(testo[-3:])      # ultimi 3 caratteri
print(testo[::2])      # un carattere ogni due: Pto
print(testo[::-1])     # al contrario: nohtyP
````

Slicing e `find` insieme permettono di "tagliare" pezzi di testo, ad esempio una email.

````python
email = "mario.rossi@esempio.it"
posizione_chiocciola = email.find("@")
utente = email[:posizione_chiocciola]
dominio = email[posizione_chiocciola + 1:]
print(utente)
print(dominio)
````
