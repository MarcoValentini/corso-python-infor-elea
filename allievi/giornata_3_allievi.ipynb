## 1. Strutture condizionali

Finora Python ha eseguito le istruzioni **tutte, in ordine**. Con una **condizione** il programma può **scegliere** quali istruzioni eseguire in base a ciò che accade (un'età, un risultato, un dato inserito dall'utente).

### 1.1 Operatori di confronto
Un confronto produce un valore `bool`: `True` (vero) o `False` (falso).

| Operatore | Significato | Esempio | Risultato |
|---|---|---|---|
| `==` | uguale | `5 == 5` | `True` |
| `!=` | diverso | `5 != 3` | `True` |
| `<` | minore | `3 < 5` | `True` |
| `>` | maggiore | `3 > 5` | `False` |
| `<=` | minore o uguale | `5 <= 5` | `True` |
| `>=` | maggiore o uguale | `4 >= 5` | `False` |

⚠️ **`=` assegna, `==` confronta.** Confondere i due è uno degli errori più comuni.

````python
x = 10
print(x == 10)
print(x != 10)
print(x < 5)
print(x >= 10)

print("Python" == "python")      # le maiuscole contano!
print("Python".lower() == "python")
````

**Confronto tra stringhe:** `==` e `!=` verificano l'uguaglianza esatta (maiuscole incluse). Con `<` e `>` si confronta l'**ordine alfabetico** (ma le maiuscole vengono prima delle minuscole: per sicurezza si usa `.lower()`).

````python
print("anna" < "bruno")
print("Zebra" < "ape")                  # True: la Z maiuscola viene prima della a minuscola!
print("Zebra".lower() < "ape".lower())  # False
````

### 1.2 `if`
```
if condizione:
    istruzioni eseguite solo se la condizione è vera
```
- I **due punti** `:` alla fine della riga sono obbligatori.
- Le istruzioni dentro l'`if` sono **indentate** (4 spazi): l'indentazione dice a Python cosa appartiene al blocco.
- Le righe non indentate vengono eseguite sempre.

````python
eta = 17

if eta >= 18:
    print("Sei maggiorenne!")
    print("Puoi avere la patente B!")

print("Fine del programma")      # eseguita sempre
````

### 1.3 `else`
`else` è la strada alternativa: viene eseguita quando la condizione dell'`if` è falsa.

````python
eta = 19

if eta >= 18:
    print("Sei maggiorenne!")
else:
    print("Sei minorenne!")
````

### 1.4 `elif`: più alternative
`elif` (*else if*) aggiunge altre condizioni. Python le controlla **in ordine** e **si ferma alla prima vera**; `else` finale raccoglie tutti gli altri casi.

````python
eta = 4

if eta >= 18:
    print("Sei maggiorenne!")
elif eta <= 6:
    print("Sei un bambino!")
else:
    print("Sei minorenne!")
````

⚠️ **L'ordine conta.** Qui sotto la prima condizione è vera già per 85, quindi le altre non vengono mai controllate: il risultato è sbagliato.

````python
voto = 85

if voto >= 60:
    print("Sufficiente")
elif voto >= 80:
    print("Ottimo")              # non verrà mai raggiunto!
````

Si ordinano quindi le condizioni dalla **più restrittiva alla meno restrittiva**:

````python
voto = 85

if voto >= 80:
    print("Ottimo")
elif voto >= 60:
    print("Sufficiente")
else:
    print("Insufficiente")
````

### 1.5 `if` annidati
Un `if` può contenere un altro `if`. Si indenta di un livello in più per ogni livello di annidamento.

````python
eta = 20
ha_patente = True

if eta >= 18:
    if ha_patente:
        print("Puoi guidare")
    else:
        print("Maggiorenne ma senza patente")
else:
    print("Sei minorenne")
````

(Spesso un `if` annidato si può semplificare con `and`: si vede più avanti.)

### 1.6 Valori booleani e condizioni salvate
Il risultato di un confronto si può salvare in una variabile e usare come condizione.

````python
eta = 17
maggiorenne = eta >= 18
print("Valore della condizione:", maggiorenne)

if maggiorenne:
    print("Sei maggiorenne!")
else:
    print("Sei minorenne!")
````

Per convenzione si scrive `if maggiorenne:` e non `if maggiorenne == True:` (è equivalente ma ridondante).

### 1.7 Valori "truthy" e "falsy"
In un `if` Python accetta **qualsiasi valore**, non solo `True` e `False`: lo converte in `bool`.

**Falsy** (contano come falso): `False`, `None`, `0`, `0.0`, `""` (stringa vuota) e le raccolte vuote che vedremo più avanti (liste, dizionari, set).
**Truthy**: **tutto il resto**, anche `"0"`, `"False"` e `" "` (stringhe non vuote) e i numeri negativi.

````python
valori = [0, 1, -1, 0.0, "", " ", "0", "False", None, 3.14]

for v in valori:
    print(f"{v!r:>8} -> {bool(v)}")
````

Un uso tipico: controllare se una stringa è vuota.

````python
nome = ""

if nome:
    print("Ciao", nome)
else:
    print("Non hai inserito il nome")
````

### 1.8 Operatori logici: `and`, `or`, `not`
Servono a combinare più condizioni.

| Operatore | Significato | Vero quando... |
|---|---|---|
| `and` | e | **entrambe** le condizioni sono vere |
| `or` | o | **almeno una** delle due è vera |
| `not` | non | inverte: vero diventa falso e viceversa |

**Tabella di verità**
| A | B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

````python
eta = 25
ha_patente = False

if eta >= 18 and ha_patente:
    print("Puoi guidare!")
else:
    print("Non puoi guidare!")

giorno = "sabato"
if giorno == "sabato" or giorno == "domenica":
    print("Siamo nel weekend!")

piove = False
if not piove:
    print("Non piove, possiamo uscire!")
````

**Precedenza:** `not` si valuta prima, poi `and`, poi `or`. Con più operatori conviene **usare le parentesi** per rendere chiaro il significato.

````python
eta = 16
ha_permesso = True
accompagnato = False

# and ha precedenza su or: questa condizione si legge come  eta >= 18  or  (ha_permesso and accompagnato)
print(eta >= 18 or ha_permesso and accompagnato)

# con le parentesi il significato cambia ed è esplicito
print((eta >= 18 or ha_permesso) and accompagnato)
````

### 1.9 Confronti concatenati
Python permette di scrivere un intervallo come in matematica, senza `and`.

````python
x = 15

if x > 10 and x < 20:
    print("Il numero è compreso tra 10 e 20.")

if 10 < x < 20:                      # equivalente e più leggibile
    print("Il numero è compreso tra 10 e 20.")
````

### 1.11 Espressione ternaria
Per assegnare uno di **due valori** in base a una condizione, si può scrivere tutto in una riga:

`variabile = valore_se_vero if condizione else valore_se_falso`

````python
ha_progetti = True
media_minima = 5.5 if ha_progetti else 6
print(media_minima)

n = 7
parita = "pari" if n % 2 == 0 else "dispari"
print(n, "è", parita)
````

Si possono concatenare, ma oltre due casi conviene usare `if/elif/else` per leggibilità.

````python
punteggio = 75
giudizio = "Ottimo" if punteggio >= 75 else "Sufficiente" if punteggio >= 60 else "Insufficiente"
print(giudizio)
````

## 2. Cicli

Un **ciclo** ripete un blocco di istruzioni più volte. Senza cicli, stampare i numeri da 1 a 1000 richiederebbe 1000 righe. Python ha due cicli: `while` (ripete **finché** una condizione è vera) e `for` (ripete **per ogni** elemento di una sequenza).

### 2.1 Il ciclo `while`
```
while condizione:
    istruzioni da ripetere
```
Python controlla la condizione: se è vera esegue il blocco e **ricontrolla**; si ferma quando diventa falsa. Per terminare, qualcosa nel blocco deve **cambiare** la situazione (di solito una variabile contatore).

````python
contatore = 1

while contatore <= 5:
    print("Giro numero", contatore)
    contatore += 1               # senza questa riga il ciclo non finirebbe mai

print("Ciclo terminato")
````

⚠️ **Ciclo infinito:** se la condizione resta sempre vera il programma non termina. Per fermarlo, in Jupyter si preme il pulsante **Interrupt** (■) della barra degli strumenti; in PyCharm il quadrato rosso **Stop**. Esempio (**non eseguire**):
```
contatore = 1
while contatore <= 5:
    print("Giro numero", contatore)     # manca contatore += 1
```
**Accumulatori:** un ciclo spesso aggiorna una variabile che "raccoglie" un risultato (somma, conteggio, prodotto).

````python
totale = 0
numero = 1

while numero <= 10:
    totale += numero
    numero += 1

print("Somma dei numeri da 1 a 10:", totale)
````

### 2.2 `while True` e `break`
`break` **interrompe subito** il ciclo. Con `while True:` (condizione sempre vera) si crea un ciclo che termina solo quando si decide, con `break`: comodo per leggere dati finché l'utente non scrive una parola di uscita.

````python
while True:
    testo = input("Scrivi un numero (stop per uscire): ")
    if testo == "stop":
        break
    print("Il doppio è", int(testo) * 2)

print("Uscito dal ciclo")
````

**Validare l'input:** si ripete la domanda finché la risposta è valida. Qui si usa `isdigit()` (vero se la stringa contiene solo cifre).

````python
while True:
    risposta = input("Password (almeno 8 caratteri): ")
    if len(risposta) >= 8:
        break
    print("Troppo corta, riprova.")

print("Password accettata")
````

### 2.3 Il ciclo `for` e `range`
`for` ripete il blocco **per ogni valore** di una sequenza. `range()` genera sequenze di numeri:

| Chiamata | Valori generati |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 (passo 2) |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 (conto alla rovescia) |

⚠️ Il valore finale è **escluso**, come negli slicing.

````python
for i in range(5):
    print(i)

print("---")
for i in range(2, 6):
    print(i)

print("---")
for i in range(0, 10, 2):
    print(i)

print("---")
for i in range(5, 0, -1):
    print(i)
````

**Quando usare `for` e quando `while`?** `for` quando si sa **quante volte** ripetere (o su quali valori); `while` quando si ripete **finché** succede qualcosa e non si sa in anticipo quanti giri serviranno.

````python
# Stesso risultato con for e con while
for i in range(1, 4):
    print("for:", i)

i = 1
while i < 4:
    print("while:", i)
    i += 1
````

La variabile del ciclo (`i`) è una normale variabile: si può usare nei calcoli.

````python
for numero in range(1, 6):
    print(f"{numero} al quadrato fa {numero ** 2}")
````

### 2.4 `break` e `continue`
- `break`: esce **del tutto** dal ciclo.
- `continue`: salta il resto del giro e passa **al successivo**.

````python
print("Con continue (salta il 3):")
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

print("Con break (si ferma al 3):")
for i in range(1, 6):
    if i == 3:
        break
    print(i)
````

### 2.5 Ciclo su una stringa
Una stringa è una sequenza di caratteri: `for` la percorre **un carattere alla volta**.

````python
for lettera in "Python":
    print(lettera)
````

````python
frase = "il volo parte alle nove"
vocali = 0

for c in frase:
    if c in "aeiou":
        vocali += 1

print("Vocali:", vocali)
````

**Costruire una stringa nuova** con un ciclo: si parte da una stringa vuota e si aggiungono caratteri.

````python
parola = "volo"
rovesciata = ""

for c in parola:
    rovesciata = c + rovesciata       # ogni lettera va davanti alle precedenti

print(rovesciata)
````

### 2.6 Cicli annidati
Un ciclo dentro un altro: per **ogni** giro del ciclo esterno, il ciclo interno gira **per intero**.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"i={i}, j={j}")
````

Uso tipico: le **tabelle** (righe e colonne). `{valore:4}` riserva 4 spazi per allineare i numeri; `print(riga)` va fuori dal ciclo interno per andare a capo a fine riga.

````python
for i in range(1, 6):
    riga = ""
    for j in range(1, 6):
        riga += f"{i * j:4}"
    print(riga)
````

### 2.7 Riepilogo del flusso
| Costrutto | Cosa fa |
|---|---|
| `while condizione:` | ripete finché la condizione è vera |
| `for x in range(...)` | ripete per ogni numero generato |
| `for c in stringa` | ripete per ogni carattere |
| `break` | esce dal ciclo |
| `continue` | passa al giro successivo |
| cicli annidati | il ciclo interno gira per intero a ogni giro dell'esterno |

## 3. Math e random

### 3.1 Funzioni matematiche di base (sempre disponibili)
| Funzione | Cosa fa | Esempio |
|---|---|---|
| `abs(x)` | valore assoluto | `abs(-3)` → `3` |
| `round(x)` | arrotonda all'intero più vicino | `round(2.7)` → `3` |
| `round(x, n)` | arrotonda a `n` decimali | `round(2.5668, 2)` → `2.57` |
| `min(a, b, ...)` / `max(a, b, ...)` | il più piccolo / il più grande | `max(3, 9, 5)` → `9` |
| `x ** n` | potenza | `2 ** 10` → `1024` |

````python
print(abs(-3), abs(3.5))
print(round(2.7), round(2.2))
print(round(2.56678, 2))
print(min(4, 9, 2), max(4, 9, 2))
````

### 3.2 L'arrotondamento "bancario"
Quando il decimale è **esattamente 5**, `round()` arrotonda al numero **pari** più vicino (arrotondamento bancario o "al pari"): serve a non introdurre un errore sistematico nei calcoli statistici.

````python
print(round(0.5), round(1.5), round(2.5), round(3.5), round(4.5))
````

Risultato: `0 2 2 4 4`. Se serve l'arrotondamento "scolastico" (per eccesso) si usano le funzioni del modulo `math` (vedi sotto).

**Attenzione ai decimali in virgola mobile:** il computer rappresenta i numeri decimali in modo approssimato, quindi a volte un caso "esatto" non lo è.

````python
print(round(2.675, 2))     # sembrerebbe 2.68, ma 2.675 in memoria è leggermente sotto: 2.67
print(0.1 + 0.2)           # 0.30000000000000004
````

### 3.3 Il modulo `math`
Le funzioni più avanzate stanno nel modulo `math`, che va **importato** una volta all'inizio.

````python
import math
````

**Arrotondamenti con `math`**
| Funzione | Cosa fa | `2.7` | `-2.7` |
|---|---|---|---|
| `math.floor(x)` | per **difetto** (verso il basso) | `2` | `-3` |
| `math.ceil(x)` | per **eccesso** (verso l'alto) | `3` | `-2` |
| `math.trunc(x)` | **tronca** la parte decimale | `2` | `-2` |

Con i numeri negativi `floor` e `trunc` differiscono. `int(x)` si comporta come `trunc`.

````python
for x in (2.7, 2.2, -2.7, -2.2):
    print(f"{x:5}: floor={math.floor(x):3}  ceil={math.ceil(x):3}  trunc={math.trunc(x):3}  int={int(x):3}")
````

**Uso tipico di `ceil`:** quando serve "quanti contenitori" per contenere un certo numero di oggetti.

````python
oggetti = 47
per_scatola = 10
scatole = math.ceil(oggetti / per_scatola)
print(scatole)       # 5: servono 5 scatole, la quinta non è piena
````

**Radice quadrata, costanti, trigonometria**
- `math.sqrt(x)` radice quadrata (restituisce sempre `float`).
- `math.pi` (π) e `math.e` (numero di Nepero).
- Le funzioni trigonometriche `math.sin`, `math.cos`, `math.tan` lavorano in **radianti**: per convertire, `math.radians(gradi)` e `math.degrees(radianti)`.

````python
print(math.sqrt(16))
print("Pi greco:", math.pi)
print("Numero di Nepero:", math.e)

angolo_gradi = 45
angolo_rad = math.radians(angolo_gradi)
print("45° in radianti:", angolo_rad)
print("seno:", math.sin(angolo_rad))
print("coseno:", math.cos(angolo_rad))
print("tangente:", math.tan(angolo_rad))        # 0.9999999999999999: errore di arrotondamento
print(round(math.tan(angolo_rad), 4))           # 1.0
````

### 3.4 Il modulo `random`: numeri casuali
Utile per giochi, simulazioni, dati di prova. Si importa con `import random`.

| Funzione | Cosa restituisce |
|---|---|
| `random.randint(a, b)` | intero casuale tra `a` e `b`, **estremi inclusi** |
| `random.uniform(a, b)` | decimale casuale tra `a` e `b` |
| `random.random()` | decimale casuale tra 0 (incluso) e 1 (escluso) |
| `random.seed(n)` | rende la sequenza **ripetibile** (utile per provare) |

````python
import random

print(random.randint(1, 6))          # lancio di un dado
print(random.uniform(10, 20))
print(random.random())
````

**Con `seed` i risultati sono sempre gli stessi** (a parità di `seed`): comodo per verificare un programma.

````python
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))      # stessi numeri
````

**Un numero casuale tra 3 e 10:** `random.randint(3, 10)`. (Si può anche fare `round(random.random() * 7) + 3`, ma è meno leggibile e non dà la stessa probabilità a tutti i valori.)

**Esempio: lanciare un dado 10 volte e sommare i risultati.**

````python
totale = 0
for lancio in range(10):
    risultato = random.randint(1, 6)
    totale += risultato
print("Somma dei 10 lanci:", totale)
````
