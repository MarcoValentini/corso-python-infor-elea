## 1. Liste
Una **lista** è una sequenza ordinata e **modificabile** di valori, scritta tra parentesi quadrate. Può contenere valori di tipi diversi, anche altre liste.

```python
lista_vuota = []
nomi = ["Anna", "Marco", "Luca", "Sara"]
numeri = [1, 2, 3]
mista = [1, "mela", 10.75, False]
matrice = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(nomi)
print(mista)
print(matrice)
```

### 1.1 Accesso e modifica
Come per le stringhe, gli elementi hanno un **indice** che parte da 0; gli indici negativi contano da destra. A differenza delle stringhe, **le liste si possono modificare** assegnando a un indice.

```python
frutta = ["mela", "pera", "banana", "arancia"]

print(frutta[0])        # primo
print(frutta[-1])       # ultimo
frutta[0] = "mandarino" # modifica
print(frutta)

matrice = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrice[0][0])    # riga 0, colonna 0
print(matrice[2][1])    # riga 2, colonna 1
```

### 1.2 Operazioni di base
| Operazione | Significato |
|---|---|
| `len(l)` | numero di elementi |
| `x in l` | `True` se `x` è nella lista |
| `l1 + l2` | unisce due liste (ne crea una nuova) |
| `l * 3` | ripete la lista 3 volte |
| `sum(l)`, `min(l)`, `max(l)` | somma, minimo, massimo (liste di numeri) |

```python
frutta = ["mela", "pera", "banana"]
print("Lunghezza:", len(frutta))
print("pera" in frutta)
print("kiwi" in frutta)
print(frutta + ["mango", "kiwi"])
print(["ciao"] * 3)

numeri = [5, 6, 10, 32, 1, 16]
print(sum(numeri), min(numeri), max(numeri))
print("Media:", sum(numeri) / len(numeri))
```

### 1.3 Scorrere una lista con `for`
Con il ciclo `for` si legge un elemento alla volta. Se serve anche la posizione si usa `enumerate()`.

```python
frutta = ["mela", "pera", "banana"]

for frutto in frutta:
    print("Frutto:", frutto)

for posizione, frutto in enumerate(frutta):
    print(posizione, frutto)

for posizione, frutto in enumerate(frutta, start=1):   # numerazione da 1
    print(f"{posizione}. {frutto}")
```

### 1.4 Metodi delle liste
⚠️ A differenza dei metodi delle stringhe, molti metodi delle liste **modificano la lista originale** e restituiscono `None`.

| Metodo | Effetto |
|---|---|
| `l.append(x)` | aggiunge `x` in fondo |
| `l.extend(altra)` | aggiunge tutti gli elementi di `altra` in fondo |
| `l.insert(i, x)` | inserisce `x` nella posizione `i` |
| `l.remove(x)` | rimuove la prima occorrenza di `x` (errore se non c'è) |
| `l.pop()` / `l.pop(i)` | rimuove e **restituisce** l'ultimo / l'elemento in posizione `i` |
| `l.sort()` / `l.sort(reverse=True)` | ordina sul posto, crescente / decrescente |
| `l.reverse()` | inverte l'ordine |
| `l.index(x)` | posizione di `x` (errore se non c'è) |
| `l.count(x)` | quante volte compare `x` |

```python
frutta = ["mela", "pera"]
frutta.append("banana")
print(frutta)

frutta.extend(["mango", "ananas"])      # append(["mango","ananas"]) aggiungerebbe UNA lista come singolo elemento
print(frutta)

frutta.insert(0, "cocco")
print(frutta)

frutta.remove("banana")
print(frutta)

ultimo = frutta.pop()
print("Rimosso:", ultimo, "->", frutta)

primo = frutta.pop(0)
print("Rimosso:", primo, "->", frutta)
```

```python
pizze = ["Margherita", "Tonno", "Quattro formaggi", "Capricciosa"]
pizze.sort()
print(pizze)
pizze.sort(reverse=True)
print(pizze)

numeri = [1, 5, 7, 2, 4]
numeri.reverse()
print(numeri)

lettere = ["a", "b", "a", "c", "a"]
print(lettere.index("b"), lettere.count("a"))
```

### 1.5 Slicing delle liste
Funziona come per le stringhe: `lista[inizio:fine:passo]`, inizio incluso e fine esclusa. Lo slicing **crea una nuova lista**.

```python
animali = ["gatto", "cane", "coniglio", "pappagallo", "pesce"]
print(animali[0:3])      # primi 3
print(animali[:3])       # equivalente
print(animali[-3:])      # ultimi 3
print(animali[::2])      # uno ogni due
print(animali[::-1])     # al contrario
```

## 2. Tuple, unpacking e copie

### 2.1 Tuple
Una **tupla** è come una lista, ma **non modificabile** (immutabile). Si scrive con le parentesi tonde. Si usa per dati che non devono cambiare, come coordinate o date.

```python
coordinate = (45.07, 7.69)
print("Latitudine:", coordinate[0])
print("Longitudine:", coordinate[1])
print(len(coordinate))

# coordinate[0] = 100   # ← TypeError: le tuple non si modificano
```

Una tupla con un solo elemento richiede la virgola: `(5,)`. Senza virgola, `(5)` è solo un numero tra parentesi.

### 2.2 Unpacking
**Unpacking** significa "spacchettare" una sequenza assegnando ogni elemento a una variabile.

```python
frutti = ("mela", "pera", "banana")
mela, pera, banana = frutti
print(mela, pera, banana)

x, y = (10, 5)
print(x, y)

nome, cognome = "Mario Rossi".split()
print(nome, cognome)

# Scambio di due valori in una riga (ricordi l'esercizio della giornata 1?)
a, b = 5, 9
a, b = b, a
print(a, b)
```

Con `*` si raccolgono più elementi in una lista. Il numero di variabili deve coincidere con gli elementi, a meno di usare `*`. Il trattino basso `_` indica "valore che non mi interessa".

```python
frutti = ["mela", "pera", "banana", "anguria", "mandarino"]

primo, secondo, *altri = frutti
print(primo, secondo, altri)

primo, *centrali, ultimo = frutti
print(primo, centrali, ultimo)

testo = "Ciao, come va? Io tutto bene"
prima, *_, ultima = testo.split()
print(prima, ultima)
```

### 2.3 Copiare le liste: attenzione al riferimento!
Scrivere `b = a` **non copia** la lista: le due variabili puntano alla **stessa** lista. Modificarne una modifica anche l'altra.

```python
animali = ["gatto", "cane", "topo"]
riferimento = animali          # NON è una copia
riferimento.append("rana")
print(animali)
print(riferimento)
```

Per una copia vera: `lista.copy()` (o `lista[:]`). Ma è una **copia superficiale**: se la lista contiene altre liste, quelle interne restano condivise. Per copiare tutto in profondità si usa `copy.deepcopy()`.

```python
import copy

animali = ["gatto", "cane", "topo"]
copia = animali.copy()
copia.append("rana")
print(animali, copia)

matrice = [[1, 2], [3, 4]]
superficiale = matrice.copy()
profonda = copy.deepcopy(matrice)

matrice[0][0] = 100
print("originale:   ", matrice)
print("superficiale:", superficiale)   # ← cambiata anche qui!
print("profonda:    ", profonda)       # ← indipendente
```

## 3. Dizionari
Un **dizionario** associa **chiavi** a **valori**: invece di accedere con una posizione (0, 1, 2…) si accede con un nome. Si scrive tra graffe, con coppie `chiave: valore`. Le chiavi sono uniche.

```python
persona = {
    "nome": "Marta",
    "cognome": "Rossi",
    "eta": 28,
}
print(persona)
print(persona["nome"])
print(persona["eta"])
vuoto = {}
```

### 3.1 Aggiungere, modificare, eliminare
- `d[chiave] = valore` aggiunge la chiave se non esiste, altrimenti ne **modifica** il valore.
- `del d[chiave]` elimina una coppia.
- Accedere a una chiave inesistente con `d[chiave]` dà un `KeyError`.

```python
persona = {"nome": "Marta", "cognome": "Rossi", "eta": 28}

persona["citta"] = "Torino"      # aggiunge
persona["eta"] = 29              # modifica
print(persona)

del persona["cognome"]           # elimina
print(persona)

# print(persona["telefono"])     # ← KeyError
```

### 3.2 Metodi dei dizionari
| Metodo | Effetto |
|---|---|
| `d.get(k)` | valore di `k`, oppure `None` se non esiste (nessun errore) |
| `d.get(k, default)` | come sopra, ma restituisce `default` se non esiste |
| `d.pop(k)` | rimuove `k` e restituisce il suo valore |
| `d.setdefault(k, v)` | se `k` non esiste la crea con valore `v`; restituisce il valore |
| `d.clear()` | svuota il dizionario |
| `len(d)` | numero di coppie |
| `k in d` | `True` se la chiave esiste (si controllano le **chiavi**) |

```python
persona = {"nome": "Marta", "eta": 28}

print(persona.get("nome"))
print(persona.get("citta"))              # None
print(persona.get("citta", "Sconosciuta"))

eta = persona.pop("eta")
print(eta, persona)

persona.setdefault("citta", "Torino")
persona.setdefault("citta", "Milano")    # esiste già: non cambia
print(persona)

print("nome" in persona, "telefono" in persona, len(persona))
```

### 3.3 Iterare sui dizionari

```python
libro = {"titolo": "I Promessi Sposi", "autore": "Alessandro Manzoni", "pagine": 544}

for chiave in libro:                      # sulle chiavi
    print(chiave)

for valore in libro.values():             # sui valori
    print(valore)

for chiave, valore in libro.items():      # su entrambi (il più usato)
    print(f"{chiave}: {valore}")
```

## 4. Set (insiemi)
Un **set** è una raccolta **senza ordine** e **senza duplicati**. Si scrive con le graffe (senza `chiave: valore`). Un set vuoto si crea con `set()` (non con `{}`, che è un dizionario vuoto).

```python
numeri = {1, 2, 3, 3, 3, 4}
print(numeri)                   # i duplicati scompaiono

lettere = ["a", "a", "b", "c", "c", "c"]
print(set(lettere))             # da lista a set: elimina i duplicati
print(list(set(lettere)))       # da set a lista (l'ordine non è garantito)

vuoto = set()
print(type(vuoto), type({}))
```

### 4.1 Metodi dei set
| Metodo | Effetto |
|---|---|
| `s.add(x)` | aggiunge un elemento |
| `s.update(iterabile)` | aggiunge più elementi |
| `s.remove(x)` | rimuove `x` (errore se non c'è) |
| `s.discard(x)` | rimuove `x` (nessun errore se non c'è) |
| `s.clear()` | svuota |
| `len(s)`, `x in s` | lunghezza, appartenenza (molto veloce) |

```python
ingredienti = {"farina", "acqua", "latte", "uova"}
ingredienti.add("burro")
ingredienti.update(["mele", "vaniglia"])
ingredienti.remove("uova")
ingredienti.discard("zucchero")   # non c'è: nessun errore
print(len(ingredienti), "farina" in ingredienti)

for ingrediente in ingredienti:   # l'ordine può cambiare a ogni esecuzione
    print("-", ingrediente)
```

### 4.2 Operazioni tra insiemi
| Operazione | Metodo | Operatore | Risultato |
|---|---|---|---|
| Unione | `a.union(b)` | `a \| b` | elementi in `a` o in `b` |
| Intersezione | `a.intersection(b)` | `a & b` | elementi in entrambi |
| Differenza | `a.difference(b)` | `a - b` | elementi in `a` ma non in `b` |
| Differenza simmetrica | `a.symmetric_difference(b)` | `a ^ b` | elementi in uno solo dei due |

```python
a = {1, 2, 3, 4, 5, 6}
b = {5, 6, 7, 8}

print("Unione:", a.union(b))
print("Intersezione:", a.intersection(b))
print("Differenza a-b:", a.difference(b))
print("Differenza simmetrica:", a.symmetric_difference(b))
print(a | b, a & b, a - b, a ^ b)
```

## 5. Gestione degli errori (`try` / `except`)
Fin qui, un errore **ferma** il programma. Con `try` / `except` possiamo prevedere l'errore e decidere cosa fare, invece di far "esplodere" il programma.

### 5.1 Errori che conosci già
| Errore | Quando succede | Esempio |
|---|---|---|
| `ValueError` | valore non convertibile | `int("ciao")` |
| `ZeroDivisionError` | divisione per zero | `10 / 0` |
| `TypeError` | tipi incompatibili | `"3" + 4` |
| `NameError` | nome non definito | `print(xyz)` |
| `IndexError` | indice fuori dai limiti | `[1, 2][5]` |
| `KeyError` | chiave inesistente | `{"a": 1}["b"]` |

### 5.2 Struttura base
Il codice nel blocco `try` viene eseguito; se sorge un errore, Python salta al blocco `except` corrispondente.

```python
try:
    numero = int(input("Inserisci un numero: "))
    print("Il numero che hai inserito è", numero)
except:
    print("Non hai inserito un numero valido!")
```

⚠️ `except:` senza tipo cattura **qualsiasi** errore e nasconde anche i bug. Meglio indicare il **tipo di errore** atteso, e si possono avere più `except`.

```python
try:
    numero = int(input("Inserisci un numero: "))
    risultato = 100 / numero
    print("100 diviso il tuo numero fa", risultato)
except ValueError:
    print("Non hai inserito un numero valido!")
except ZeroDivisionError:
    print("Non puoi dividere per zero!")
```

### 5.3 Leggere il messaggio dell'errore: `as`

```python
try:
    numero = int("ciao")
except ValueError as errore:
    print("Si è verificato un errore:")
    print(errore)
```

### 5.4 `else` e `finally`
- `else`: eseguito **solo se non ci sono stati errori**.
- `finally`: eseguito **sempre**, errore o no (utile per chiusure e messaggi finali).

```python
try:
    numero = int(input("Inserisci un numero: "))
except ValueError:
    print("Numero non valido")
else:
    print("Nessun errore, hai scritto", numero)
finally:
    print("Questo viene stampato SEMPRE.")
```

### 5.5 Pattern fondamentale: ripetere finché l'input è valido
Si combina `while True` con `try`/`except`: si esce dal ciclo con `break` solo quando la conversione riesce.

```python
while True:
    try:
        eta = int(input("Quanti anni hai? "))
        break                      # conversione riuscita: esco dal ciclo
    except ValueError:
        print("Valore non valido, riprova.")

print("Hai", eta, "anni")
```

## 6. Il debugger di PyCharm
Quando un programma non fa ciò che ti aspetti, aggiungere tanti `print()` funziona, ma è lento. Il **debugger** permette di **fermare** il programma su una riga e osservarne lo stato passo per passo.

> Il debugger si usa su un file `.py` in PyCharm. Crea un nuovo file `debug_prova.py` nel progetto e copia dentro il codice dell'esercizio.

### 6.1 Concetti
| Termine | Significato |
|---|---|
| **Breakpoint** | segnaposto (pallino rosso) su una riga: il programma si ferma **prima** di eseguirla |
| **Step Over** | esegue la riga corrente e passa alla successiva |
| **Step Into** | entra dentro una funzione chiamata nella riga corrente (la useremo con le funzioni) |
| **Resume** | continua fino al prossimo breakpoint |
| **Variables** | pannello che mostra i valori attuali di tutte le variabili |
| **Watches** | espressioni a scelta (es. `somma / 2`) valutate a ogni passo |

### 6.2 Procedura (Windows/Linux; su macOS le scorciatoie possono cambiare)
1. Clicca nel margine sinistro accanto al numero di riga: compare un **pallino rosso**.
2. Avvia in modalità debug: tasto destro sul file → **Debug**, oppure l'icona dell'insetto (scorciatoia `Shift + F9`).
3. Il programma si ferma sulla riga evidenziata. Guarda il pannello **Variables**.
4. Avanza riga per riga con **Step Over** (`F8`). Dopo ogni passo controlla come cambiano le variabili.
5. Per terminare, **Resume** (`F9`) o il quadrato rosso **Stop**.

### 6.3 Esempio guidato
Questo programma dovrebbe calcolare la media dei voti, ma stampa un valore sbagliato. **Senza modificarlo**, mettilo in un file `.py`, metti un breakpoint sulla riga `somma = v` e avanza con Step Over osservando `somma` a ogni giro del ciclo.

```python
voti = [6, 8, 7, 9]
somma = 0
for v in voti:
    somma = v
media = somma / len(voti)
print("Media:", media)
```

**Cosa osservi?** Alla fine del ciclo `somma` vale 9 (l'ultimo voto), non 30. Il codice **sostituisce** `somma` a ogni giro invece di **accumularla**. Correzione: `somma += v`.

#### 🧪 Esercizio – Caccia al bug con il debugger  ⭐
