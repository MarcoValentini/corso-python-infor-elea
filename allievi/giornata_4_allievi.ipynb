## 1. Comprehension

Una **comprehension** è un modo compatto per costruire una lista (o un set, o un dizionario) a partire da un'altra sequenza, in una sola riga. Prima si vede la versione con il ciclo, poi quella compatta.

### 1.1 List comprehension
Schema: `[espressione for elemento in sequenza]`

```python
numeri = [2, 3, 4, 5, 6]

# Versione con ciclo
quadrati = []
for n in numeri:
    quadrati.append(n ** 2)
print(quadrati)

# Versione con list comprehension
quadrati = [n ** 2 for n in numeri]
print(quadrati)
```

```python
citta = ["Torino", "Milano", "Roma"]
sigle = [c[:3].upper() for c in citta]
print(sigle)
```

### 1.2 Con una condizione (filtro)
Schema: `[espressione for elemento in sequenza if condizione]`: tiene solo gli elementi che soddisfano la condizione.

```python
temperature = [22.5, 24, 25.1, 26, 23.3, 20.2, 25.5]

calde = [t for t in temperature if t > 25]
print(calde)
```

### 1.3 Con un'espressione condizionale (ternario)
Schema: `[valore_se_vero if condizione else valore_se_falso for elemento in sequenza]`: **trasforma** ogni elemento in uno di due valori, senza scartarne nessuno.

⚠️ Differenza da ricordare: `if` **in fondo** = filtro (alcuni elementi spariscono); `if ... else` **all'inizio** = scelta del valore (nessun elemento sparisce).

```python
etichette = ["caldo" if t > 25 else "freddo" for t in temperature]
print(etichette)
```

### 1.4 Set comprehension
Come la list comprehension ma con le graffe: il risultato non ha duplicati.

```python
lettere = ["A", "a", "B", "c", "C", "c", "b", "d"]
uniche = {l.lower() for l in lettere}
print(uniche)
```

### 1.5 Dictionary comprehension
Schema: `{chiave: valore for elemento in sequenza}`

```python
numeri = [2, 3, 4]
quadrati = {f"quadrato_di_{n}": n ** 2 for n in numeri}
print(quadrati)

prezzi = {"pane": 1.5, "pizza": 6.5, "piadina": 4.5}
costosi = {prodotto: prezzo for prodotto, prezzo in prezzi.items() if prezzo > 4}
print(costosi)
```

## 2. Funzioni

Una **funzione** è un blocco di codice con un nome, scritto una volta e riutilizzabile quante volte serve. Serve a evitare ripetizioni e a dividere un programma in parti piccole e leggibili.

**Due fasi:** si **definisce** con `def` (e non succede nulla), si **chiama** scrivendo il nome seguito da `()` (e il codice viene eseguito). Il corpo è **indentato** (4 spazi). I nomi seguono `snake_case` e dovrebbero dire cosa fa la funzione.

```python
def saluta():
    print("Ciao!")

saluta()
saluta()
```

### 2.1 Parametri
I **parametri** sono le variabili elencate nella definizione; gli **argomenti** sono i valori passati quando si chiama la funzione.

```python
def saluta(nome):               # nome è il parametro
    print(f"Ciao {nome}!")

saluta("Marta")                 # "Marta" è l'argomento
saluta("Luca")

def presenta(nome, cognome):
    print(f"Ciao {nome} {cognome}!")

presenta("Marta", "Rossi")
```

**Argomenti posizionali e nominati.** Per posizione conta l'ordine; con il nome (`parametro=valore`) l'ordine è libero. In una chiamata gli argomenti posizionali vanno **prima** di quelli nominati.

```python
def descrivi(nome, eta, citta):
    print(f"{nome}, {eta} anni, vive a {citta}.")

descrivi("Anna", 28, "Torino")                  # posizionali
descrivi(eta=28, citta="Torino", nome="Anna")   # nominati, ordine libero
descrivi("Anna", citta="Torino", eta=28)        # misti
```

### 2.2 Valore di ritorno: `return`
`print` **mostra** un valore; `return` lo **restituisce** al codice che ha chiamato la funzione, che può salvarlo in una variabile o usarlo in un calcolo. Quando Python incontra `return`, la funzione termina subito. Senza `return` la funzione restituisce `None`.

```python
def somma(a, b):
    risultato = a + b
    return risultato

totale = somma(10, 9)
print(totale)
print(somma(1, 2) * 10)      # il valore restituito si usa in un calcolo

def senza_return(a):
    a + 1                    # calcola ma non restituisce nulla

print(senza_return(5))       # None
```

### 2.3 Restituire più valori
Si separano i valori con virgole: Python li raggruppa in una **tupla**. Con l'**unpacking** si assegnano a variabili distinte.

```python
def operazioni(a, b):
    return a + b, a - b, a * b, a / b

somma_, differenza, prodotto, quoziente = operazioni(10, 4)
print(somma_, differenza, prodotto, quoziente)

risultato = operazioni(10, 4)       # senza unpacking: una tupla
print(risultato)
```

### 2.4 Valori predefiniti (default)
Un parametro può avere un valore di **default**, usato se la chiamata non lo specifica. I parametri con default vanno **dopo** quelli obbligatori.

## 3. `*args` e `**kwargs`: numero variabile di argomenti

A volte non si sa in anticipo quanti argomenti arriveranno.

### 3.1 `*args`: argomenti posizionali in numero qualsiasi
Il nome preceduto da `*` raccoglie tutti gli argomenti posizionali in una **tupla**. (`args` è solo una convenzione.)

```python
def somma_tutti(*numeri):
    totale = 0
    for n in numeri:
        totale += n
    return totale

print(somma_tutti(1, 2, 3))
print(somma_tutti(1, 2, 3, 4, 5, 6))
print(somma_tutti())
```

### 3.2 `**kwargs`: argomenti nominati in numero qualsiasi
Il nome preceduto da `**` raccoglie gli argomenti nominati in un **dizionario**. (`kwargs` = *keyword arguments*.)

```python
def stampa_scheda(**dati):
    print(" Scheda ".center(30, "="))
    for chiave, valore in dati.items():
        print(f"{chiave}: {valore}")
    print("=" * 30)

stampa_scheda(nome="Mario", anni=20, citta="Roma")
```

### 3.3 Unpacking nella chiamata
Con `*` davanti a una lista/tupla e `**` davanti a un dizionario, nella **chiamata** si "spacchettano" i valori come argomenti separati.

```python
valori = [4, 5, 6]
print(somma_tutti(*valori))          # come somma_tutti(4, 5, 6)

info = {"nome": "Giulia", "anni": 31}
stampa_scheda(**info)                # come stampa_scheda(nome="Giulia", anni=31)
```

### 3.4 Combinare i tipi di parametri
L'ordine nella definizione è sempre: **obbligatori → default → `*args` → `**kwargs`**.

```python
def crea_menu(ristorante, *piatti, **dettagli):
    print("=" * 30)
    print(f"Menù di {ristorante}")
    print("=" * 30)
    for piatto in piatti:
        print("-", piatto)
    print("=" * 30)
    for chiave, valore in dettagli.items():
        print(f"{chiave}: {valore}")

crea_menu("Da Mario", "Pizza margherita", "Pasta al sugo", chef="Mario", orario="19:00-22:00")
```

## 4. Scope: dove vive una variabile

Lo **scope** (ambito) è la zona del programma in cui una variabile esiste.
- **Locale**: creata dentro una funzione, esiste solo lì dentro e sparisce al termine.
- **Globale**: creata fuori da qualsiasi funzione, leggibile ovunque.

```python
nome = "Anna"                      # globale

def saluta():
    print("Nome globale:", nome)   # legge la globale

saluta()

def saluta_locale():
    nome_locale = "Luca"           # locale
    print("Nome locale:", nome_locale)

saluta_locale()
# print(nome_locale)               # NameError: non esiste fuori dalla funzione
```

**Assegnare dentro una funzione crea una variabile locale** con lo stesso nome, senza toccare quella globale.

```python
nome = "Anna"

def cambia():
    nome = "Luca"                  # nuova variabile locale!
    print("Dentro:", nome)

cambia()
print("Fuori:", nome)              # ancora Anna
```

La parola `global` permette di modificare la variabile globale dalla funzione.

⚠️ Nella pratica è **meglio evitarla**: è più chiaro passare valori come parametri e riceverli con `return`.

```python
contatore = 0

def incrementa_globale():
    global contatore
    contatore += 1

incrementa_globale()
incrementa_globale()
print(contatore)                   # 2

# Versione consigliata: niente global, si passa e si restituisce
def incrementa(valore):
    return valore + 1

contatore = 0
contatore = incrementa(contatore)
contatore = incrementa(contatore)
print(contatore)                   # 2
```

**(facoltativo) `nonlocal`:** serve in funzioni annidate per modificare una variabile della funzione esterna.

```python
def esterna():
    totale = 0
    def interna():
        nonlocal totale
        totale += 10
    interna()
    interna()
    return totale

print(esterna())    # 20
```

## 5. Funzioni lambda, `map` e `filter`

### 5.1 Lambda
Una **funzione lambda** è una funzione senza nome, scritta in una riga, con una sola espressione il cui risultato viene restituito da sé. Sintassi: `lambda parametri: espressione`.

```python
def quadrato(x):
    return x ** 2

quadrato_lambda = lambda x: x ** 2      # equivalente

print(quadrato(5), quadrato_lambda(5))

somma = lambda a, b: a + b
print(somma(1, 2))

massimo = lambda a, b: a if a > b else b    # con espressione condizionale
print(massimo(10, 20))
```

Le lambda sono utili soprattutto **al volo**, come argomenti di altre funzioni.

### 5.2 `map`: applica una funzione a ogni elemento
`map(funzione, sequenza)` restituisce un oggetto "pigro": si converte in lista con `list()`.

```python
numeri = [1, 2, 3, 4, 5, 6]
quadrati = list(map(lambda x: x ** 2, numeri))
print(quadrati)

def quadrato(x):
    return x ** 2
print(list(map(quadrato, numeri)))      # anche con una funzione normale
```

### 5.3 `filter`: tiene solo gli elementi che soddisfano una condizione
`filter(funzione, sequenza)` tiene gli elementi per cui la funzione restituisce `True`.

```python
numeri = list(range(1, 11))
pari = list(filter(lambda x: x % 2 == 0, numeri))
print(pari)
```

**Quale scegliere?** `map` e `filter` fanno ciò che fanno le comprehension. Per lo stesso risultato: `[x ** 2 for x in numeri]` e `[x for x in numeri if x % 2 == 0]`. Le comprehension sono di solito più leggibili; `map` e `filter` vanno conosciuti perché si trovano spesso nel codice altrui.

### 5.4 `sorted` con una lambda come criterio
Il parametro `key` dice **secondo cosa** ordinare.

```python
voli = [("AZ101", 120), ("LH202", 85), ("AF303", 150)]

per_passeggeri = sorted(voli, key=lambda volo: volo[1])
print(per_passeggeri)

decrescente = sorted(voli, key=lambda volo: volo[1], reverse=True)
print(decrescente)

parole = ["elica", "ala", "carrello", "flap"]
print(sorted(parole, key=lambda p: len(p)))
```

### 5.5 (facoltativo) `reduce`
`reduce(funzione, sequenza)` combina gli elementi a due a due fino a ottenerne uno solo. Va importata da `functools`.

```python
from functools import reduce

numeri = [1, 2, 3, 4, 5]
somma = reduce(lambda acc, x: acc + x, numeri)
prodotto = reduce(lambda acc, x: acc * x, numeri)
print(somma, prodotto)
```

## 6. Moduli: il proprio file di funzioni

Un **modulo** è un file `.py` che contiene funzioni (e variabili) riutilizzabili da altri file. Si importa con `import`. Finora abbiamo usato moduli già pronti (`math`, `random`): ora ne scriviamo uno nostro.

### 6.1 Creare il modulo `utils.py`
Nella stessa cartella del notebook crea un file `utils.py`. Per comodità la cella seguente lo crea da sola (in PyCharm puoi anche crearlo con *New → Python File* e incollare il contenuto).

```python
%%writefile utils.py
NODO_IN_KMH = 1.852

def nodi_in_kmh(nodi):
    """Converte nodi in km/h."""
    return nodi * NODO_IN_KMH

def piedi_in_metri(piedi):
    """Converte piedi in metri."""
    return piedi * 0.3048

def media(*valori):
    """Media dei valori ricevuti."""
    return sum(valori) / len(valori)
```

### 6.2 Importare e usare
Tre modi, tutti equivalenti nel risultato:

```python
import utils
print(utils.nodi_in_kmh(250))
print(utils.NODO_IN_KMH)
```

```python
from utils import piedi_in_metri, media
print(piedi_in_metri(35000))
print(media(6, 8, 7))
```

```python
import utils as u          # alias più breve
print(u.media(10, 20, 30))
```

- `import modulo` → si usa `modulo.funzione()` (chiaro da dove arriva).
- `from modulo import funzione` → si usa `funzione()` direttamente.
- `import modulo as alias` → nome abbreviato (si usa spesso con le librerie, come vedrai in seguito).

⚠️ **Non chiamare mai un tuo file come un modulo esistente** (`random.py`, `math.py`): Python importerebbe il tuo file al posto di quello vero e darebbe errori strani.

Se modifichi `utils.py` mentre il notebook è aperto, riavvia il kernel (o usa `importlib.reload`) per vedere le modifiche.
